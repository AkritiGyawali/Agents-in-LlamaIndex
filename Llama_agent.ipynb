```python
import asyncio
import nest_asyncio
nest_asyncio.apply()

from google.colab import userdata

from llama_index.llms.openai import OpenAI
from llama_index.core.agent.workflow import AgentWorkflow
from llama_index.core.tools import FunctionTool
from llama_index.core.workflow import Context # Moved this up for clarity

HF_TOKEN = userdata.get("Ch2")

#define simple tool
def multiply(a:int,b:int) -> int:
  """Multiply two integers and return the integer"""
  return a*b


#initilize models
llm = OpenAI(
    model_id = 'openrouter/free',
    api_base="https://openrouter.ai/api/v1",
    api_key=HF_TOKEN
)

# initialize agent
agent = AgentWorkflow.from_tools_or_functions(
    [FunctionTool.from_defaults(multiply)],
    llm = llm
)

async def main():
    # Stateless interaction
    print("[1m Stateless Interaction ")
    response_stateless = await agent.run("What is 2 times 2?")
    print(f"Stateless Response: {response_stateless}")

    # Remembering state interaction
    print("\n[1m Stateful Interaction ")
    ctx = Context(agent)
    # Changed 'run_async' to 'run'
    response_bob = await agent.run("My name is Bob." , ctx = ctx)
    print(f"Response after 'My name is Bob.': {response_bob}")
    # Changed 'run_async' to 'run'
    response_name_again = await agent.run("What was my name again" , ctx=ctx)
    print(f"Response after 'What was my name again': {response_name_again}")

# Run the async main function
asyncio.run(main())
```
